```python
t_ref, vx_ref, vy_ref, w_ref = load_log("vel_ref.log")
t_real, vx_real, vy_real, w_real = load_log("vel_real.log")

```

```python
plt.figure()
plt.plot(t_ref, vx_ref, label="vx ref")
plt.plot(t_real, vx_real, '--', label="vx real")
plt.xlabel("Tiempo [s]")
plt.ylabel("Velocidad [m/s]")
plt.title("Velocidad lineal x")
plt.legend()
plt.grid()
plt.show()
```
